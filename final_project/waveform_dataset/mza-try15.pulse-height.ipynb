```python
# mostly pulled from hls4ml_3.ipynb and waveform_demo.ipynb
# https://fastmachinelearning.org/hls4ml/api/profiling.html
# last updated 2025-05-15 by mza
name = "mza_try15"
num_epochs = 10000
batch_size = 1000
truths_to_use = [ 2 ] # t_peak, t_sigma, height, pedestal
hidden_nodes = [ 14 ]; learning_rate = 0.1; hidden_node_type = "relu"
#hidden_nodes = [ 224 ]; learning_rate = 0.0825; hidden_node_type = "relu" # keras_tuner suggests: {'num_layers': 1, 'units_0': 224, 'activation': 'relu', 'lr': 0.08249331566936491}
#hidden_nodes = [ 232, 240 ]; learning_rate = 0.0825; hidden_node_type = "relu" # keras_tuner suggests: {'num_layers': 2, 'units_0': 232, 'activation': 'relu', 'lr': 0.08249331566936491, 'units_1': 240}
#hidden_nodes = [ 488, 240, 432 ]; learning_rate = 0.0715; hidden_node_type = "relu" # keras_tuner suggests: {'num_layers': 3, 'units_0': 488, 'activation': 'relu', 'lr': 0.07147424082020044, 'units_1': 240, 'units_2': 432}
model_reusefactor = 14 # Valid ReuseFactor(s): 1,2,4,7,8,14,16,28,32,56,112,224
input_qdense_reusefactor = 1000 # Valid ReuseFactor(s): 1,2,4,5,10,20,25,50,100,200,400,500,1000,2000,2500,5000,10000
hidden0qdense_reusefactor = 1400 # Valid ReuseFactor(s): 1,2,4,5,10,20,25,50,100,200,700,1400
use_pruning = False
prune_ratio = 0.1
train = True
synth_hls = True
# input quantization:
input_kernel_quantization = 7
input_kernel_integer_part = 0
input_bias_quantization = 9
input_bias_integer_part = 0
input_activation_quantization = 7
# hidden layer quantization:
inner_kernel_quantization = 7
inner_kernel_integer_part = 0
inner_bias_quantization = 9
inner_bias_integer_part = 0
inner_activation_quantization = 7
# output quantization:
output_kernel_quantization = 7
output_kernel_integer_part = 3
output_bias_quantization = 9
output_bias_integer_part = 0
output_cartoon_quantization = 7
output_cartoon_integer_part = 3
default_quantization = 7
default_integer_part = 3
csim = False
cosim = False
```

```python
import os, errno
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
os.environ['PATH'] = os.environ['XILINX_VIVADO'] + '/bin:' + os.environ['PATH']
import numpy as np
seed = 0
np.random.seed(seed)
import matplotlib.pyplot as plt
try:
    os.mkdir(name)
except OSError as exception:
    if exception.errno != errno.EEXIST:
        raise
import tensorflow as tf
tf.random.set_seed(seed)
from tensorflow.keras.models import Sequential
from tensorflow.keras.regularizers import l1
from tensorflow.keras.layers import Activation
from tensorflow.keras import optimizers
from callbacks import all_callbacks
from qkeras.qlayers import QDense, QActivation
from qkeras.quantizers import quantized_bits, quantized_relu, quantized_sigmoid
```

```python
dataset1 = np.load('waveform_data_0.npy') # (10000, 104)
dataset2 = np.load('waveform_data_1.npy') # (10000, 104)
dataset = np.concatenate((dataset1, dataset2))
num_waveforms = len(dataset)
training_quantity = int(0.8 * num_waveforms)
num_truths = 4
time_samples = len(dataset[0]) - num_truths
#print("num_waveforms: " + str(num_waveforms))
num_truths_to_use = len(truths_to_use)
#print("num_truths_to_use: " + str(num_truths_to_use))
waveforms = dataset[:,num_truths:]
truths = dataset[:,:num_truths]
waveform_min = min([ min(waveforms[i]) for i in range(len(waveforms)) ])
waveform_max = max([ max(waveforms[i]) for i in range(len(waveforms)) ])
offset = waveform_min
gain = 1.0 / (waveform_max - waveform_min)
scaled_waveforms = np.array([ [ gain * (waveforms[j,i] - offset) for i in range(time_samples) ] for j in range(num_waveforms) ])
scaled_truths = np.array([ [ truths[j,0]/time_samples, truths[j,1]/time_samples, gain * (truths[j,2] - offset), gain * (truths[j,3] - offset) ] for j in range(num_waveforms) ])
train_data = scaled_waveforms[:training_quantity,:]
#print("train_data.shape: " + str(train_data.shape))
train_truth = scaled_truths[:training_quantity,truths_to_use]
#print("train_truth.shape: " + str(train_truth.shape))
test_data = scaled_waveforms[training_quantity:,:]
#print("test_data.shape: " + str(test_data.shape))
test_truth = scaled_truths[training_quantity:,truths_to_use]
#print("test_truth.shape: " + str(test_truth.shape))
```

```python
evt = 8
if 0:
    x = np.zeros(time_samples)
    for i in range(time_samples):
        x[i] = i/time_samples
    y = train_data[evt]
    plt.xlim(0, 1), plt.ylim(0, 1)
    plt.scatter(x, y)
    plt.plot([0.0, 1.0], [scaled_truths[evt][3] + scaled_truths[evt][2], scaled_truths[evt][3] + scaled_truths[evt][2]], color="green")
    plt.show()
```

```python
input_kernel_quantizer = quantized_bits(bits=input_kernel_quantization, integer=input_kernel_integer_part, alpha=1)
input_bias_quantizer = quantized_bits(bits=input_bias_quantization, integer=input_bias_integer_part, alpha=1)
inner_kernel_quantizer = quantized_bits(bits=inner_kernel_quantization, integer=inner_kernel_integer_part, alpha=1)
inner_bias_quantizer = quantized_bits(bits=inner_bias_quantization, integer=inner_bias_integer_part, alpha=1)
output_kernel_quantizer = quantized_bits(bits=output_kernel_quantization, integer=output_kernel_integer_part, alpha=1)
output_bias_quantizer = quantized_bits(bits=output_bias_quantization, integer=output_bias_integer_part, alpha=1)
Q = "RND_CONV"; O = "SAT"; N = "0"
default_quantization_string = 'ap_fixed<' + str(default_quantization) + ',' + str(default_integer_part) + ',' + Q + ',' + O + ',' + N + '>'
output_quantization_string = 'ap_fixed<' + str(output_cartoon_quantization) + ',' + str(output_cartoon_integer_part) + ',' + Q + ',' + O + ',' + N + '>'
if hidden_node_type=="relu":
    hidden_activations = [ quantized_relu(bits=inner_activation_quantization) for i in range(len(hidden_nodes)) ]
else:
    hidden_activations = [ quantized_sigmoid(bits=inner_activation_quantization) for i in range(len(hidden_nodes)) ]
hidden_names = [ hidden_node_type for i in range(len(hidden_nodes)) ]
model = Sequential()
model.add(QDense(units=time_samples, input_shape=(time_samples,), name='input_qdense', kernel_quantizer=input_kernel_quantizer, bias_quantizer=input_bias_quantizer, kernel_initializer='glorot_uniform'))
model.add(QActivation(activation=quantized_sigmoid(bits=input_activation_quantization), name='input_sigmoid'))
for i in range(len(hidden_nodes)):
    model.add(QDense(units=hidden_nodes[i], name=hidden_names[i] + str(i) + "qdense", kernel_quantizer=inner_kernel_quantizer, bias_quantizer=inner_bias_quantizer, kernel_initializer='glorot_uniform'))
    model.add(QActivation(activation=hidden_activations[i], name=hidden_names[i] + str(i) + "activation"))
model.add(QDense(units=num_truths_to_use, name='output_qdense', kernel_quantizer=output_kernel_quantizer, bias_quantizer=output_bias_quantizer, kernel_initializer='glorot_uniform'))
if num_truths_to_use<2:
    model.add(Activation(activation='sigmoid', name='output_sigmoid'))
    output_layer_name = 'output_sigmoid'
else:
    model.add(Activation(activation='softmax', name='output_softmax'))
    output_layer_name = 'output_softmax'
```

```python
import callbacks # https://keras.io/guides/writing_your_own_callbacks/
class CustomCallback(callbacks.Callback):
    def on_epoch_begin(self, epoch, logs=None):
        if epoch%1000==0:
            print("epoch: " + str(epoch) + "/" + str(num_epochs))
        if epoch==num_epochs-1:
            print("epoch: " + str(epoch) + "/" + str(num_epochs))
```

```python
if use_pruning:
    from tensorflow_model_optimization.python.core.sparsity.keras import prune, pruning_callbacks, pruning_schedule
    from tensorflow_model_optimization.sparsity.keras import strip_pruning
    # the first argument is the percentage of the weights that will be forced to be 0
    pruning_params = {"pruning_schedule": pruning_schedule.ConstantSparsity(prune_ratio, begin_step=2000, frequency=100)}
    model = prune.prune_low_magnitude(model, **pruning_params)
```

```python
if train:
    #optimizer = optimizers.Adam(learning_rate=learning_rate)
    optimizer = optimizers.SGD(learning_rate=learning_rate)
    #model.compile(optimizer=optimizer, loss=['categorical_crossentropy'], metrics=['accuracy'])
    model.compile(optimizer=optimizer, loss=['mse'])
    if use_pruning:
        callbacks = all_callbacks(stop_patience=1000, lr_factor=0.5, lr_patience=10, lr_epsilon=0.000001, lr_cooldown=2, lr_minimum=0.0000001, outputDir=name)
        callbacks.callbacks.append(pruning_callbacks.UpdatePruningStep())
        model.fit(train_data, train_truth, batch_size=batch_size, epochs=num_epochs, validation_split=0.25, shuffle=False, callbacks=callbacks.callbacks, verbose=0)
    else:
        model.fit(train_data, train_truth, batch_size=batch_size, epochs=num_epochs, validation_split=0.25, shuffle=False, verbose=0, callbacks=[CustomCallback()])
    #loss, acc = model.evaluate(test_data, test_truth, verbose=0)
    if use_pruning:
        model = strip_pruning(model)
    model.save(name + '/KERAS_check_best_model.keras')
else:
    from tensorflow.keras.models import load_model
    from qkeras.utils import _add_supported_quantized_objects
    co = {}
    _add_supported_quantized_objects(co)
    model = load_model(name + '/KERAS_check_best_model.keras', custom_objects=co) # TypeError: <qkeras.qlayers.QActivation object at 0x774300567100> could not be deserialized properly. Please ensure that components that are Python object instances (layers, models, etc.) returned by `get_config()` are explicitly deserialized in the model's `from_config()` method.
```

```python
#print('test_acc: {:.3f}, test_loss: {:.3f}'.format(acc, loss))
train_prediction = model.predict(train_data)
test_prediction = model.predict(test_data)
```

```python
import hls4ml, plotting
config = hls4ml.utils.config_from_keras_model(model, granularity='name')
config['LayerName'][output_layer_name]['exp_table_t'] = output_quantization_string
config['LayerName'][output_layer_name]['inv_table_t'] = output_quantization_string
config['LayerName'][output_layer_name]['Precision']   = output_quantization_string
config['IOType'] = 'io_stream' # io_parallel/io_stream
config['Model']['Precision']['default'] = default_quantization_string
config['Model']['Strategy'] = 'Resource'
config['Model']['ReuseFactor'] = model_reusefactor
config['LayerName']['input_qdense']['ReuseFactor'] = input_qdense_reusefactor
config['LayerName'][hidden_names[0] + "0qdense"]['ReuseFactor'] = hidden0qdense_reusefactor
config['Model']['TraceOutput'] = True
for layer in config['LayerName'].keys():
    config['LayerName'][layer]['Trace'] = True
    config['LayerName'][layer]['Strategy'] = 'Resource'
plotting.print_dict(config)
hls_model = hls4ml.converters.convert_from_keras_model(
    model, hls_config=config, output_dir=name + '/hls4ml_prj', part='xc7z020clg400-1',
    input_data_tb = name + "/Input.dat", output_data_tb = name + "/Output.dat", backend='Vitis', verbose=0)
```
